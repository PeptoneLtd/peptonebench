```python
import pandas as pd
import numpy as np
import json
import os

df = pd.read_csv(
    "../../datasets/PeptoneDB-Integrative/PeptoneDB-Integrative.csv",
    converters={"gscores": lambda x: np.array(json.loads(x))},
    index_col="label",
)

fasta_dir = "../predictions/fasta/"
os.makedirs(fasta_dir, exist_ok=True)
for label, row in df.iterrows():
    with open(os.path.join(fasta_dir, f"{label}.fasta"), "w") as f:
        f.write(">" + label + "\n" + row["sequence"] + "\n")

gscore_root = "../gscores/"
os.makedirs(gscore_root, exist_ok=True)
for label, row in df.iterrows():
    os.makedirs(os.path.join(gscore_root, f"{label}"), exist_ok=True)
    with open(os.path.join(gscore_root, f"{label}/adopt2_gscores.csv"), "w") as f:
        f.write("residue,gscore\n")
        seq = row["sequence"]
        for i, gscore in enumerate(row["gscores"]):
            f.write(f"{seq[i]},{gscore}\n")
```
